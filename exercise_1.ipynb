# Data Science - numpy: exercise_1

```python
import numpy as np
```

### 1. Creating arrays with realistic data

```python
print("Creating an array with test scores:")
test_scores = np.array([85, 92, 78, 65, 98, 89, 75])
print(test_scores)
```

### 2. Creating a 2D array with temperature data

```python
print("\nWeekly temperatures (Mon-Fri,°C):")
weekly_temps = np.array([
    [32, 25, 28, 31, 19],  # Monday-Friday, Week 1
    [20, 22, 25, 29, 25]
    ])   # Monday-Friday, Week 2
print(weekly_temps)
```

### 3. Statistical operations on test scores

```python
print(f"Sum of all scores: {np.sum(test_scores)}")
print(f"Average score: {np.mean(test_scores)}")
print(f"Highest score: {np.max(test_scores)}")
print(f"Lowest score: {np.min(test_scores)}")
print(f"Standard deviation: {np.std(test_scores)}")
print(f"Median score: {np.median(test_scores)}")
```

### 4. Indexing and Slicing

```python
print(f"First test score: {test_scores[0]}")
print(f"First three test scores: {test_scores[0:3]}")
print(f"Week 1 temperatures: {weekly_temps[0]}")
print(f"Friday temperature in Week 2: {weekly_temps[1, 4]}")
print(f"Wednesday temperature in Week 2: {weekly_temps[1][2]}")
```

### 5. Creating a random array of integers between 20 and 55 (inclusive), size = 10

```python
arr = np.random.randint(20,55, size=10 )
print(arr)
```
